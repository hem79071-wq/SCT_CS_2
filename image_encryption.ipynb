```python
from PIL import Image
import numpy as np

# Load image
img = Image.open("input.png")
img = img.convert("RGB")

# Convert to numpy array (int type)
data = np.array(img, dtype=int)

# Key
key = 50

# Encryption
encrypted_data = (data + key) % 256

# Save encrypted image
encrypted_img = Image.fromarray(encrypted_data.astype('uint8'))
encrypted_img.save("encrypted.png")

print("✅ Image Encrypted Successfully")
```

```python
from PIL import Image
import numpy as np
from IPython.display import display

# Load encrypted image
encrypted_img = Image.open("encrypted.png")
encrypted_img = encrypted_img.convert("RGB")

# Convert to numpy array (int type)
encrypted_data = np.array(encrypted_img, dtype=int)

# Same key used for encryption
key = 50

# DECRYPTION
decrypted_data = (encrypted_data - key) % 256

# Convert back to image and save
decrypted_img = Image.fromarray(decrypted_data.astype('uint8'))
decrypted_img.save("decrypted.png")

print("✅ Image Decrypted Successfully")

# Display decrypted image
display(decrypted_img)
```

```python
from PIL import Image
import numpy as np

# Load original image
img = Image.open("input.png")
img = img.convert("RGB")

# Convert to numpy array
data = np.array(img, dtype=int)

# Key for encryption
key = 50

# ENCRYPTION
encrypted_data = (data + key) % 256
encrypted_img = Image.fromarray(encrypted_data.astype('uint8'))

# Save encrypted image
encrypted_img.save("encrypted.png")
print("✅ Encrypted image created successfully")
```
